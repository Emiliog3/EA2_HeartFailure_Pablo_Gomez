## 1. Importaciones

Primero reunimos las herramientas mínimas. `datasets` permite solicitar el conjunto publicado en Hugging Face; `pandas` administra el DataFrame; `Path` ayuda a construir rutas portables; y `display` muestra tablas con mejor formato dentro de Jupyter.

```python
# Cómo se lee: “De la librería datasets importamos la función load_dataset”.
# Para qué sirve: permite obtener directamente un dataset publicado en Hugging Face.
from datasets import load_dataset

# Cómo se lee: “Importamos pandas y le asignamos el alias pd”.
# Para qué sirve: permite trabajar con el conjunto como una tabla o DataFrame.
import pandas as pd

# Cómo se lee: “De pathlib importamos la clase Path”.
# Para qué sirve: ayuda a construir carpetas y rutas sin depender de separadores manuales.
from pathlib import Path

# Cómo se lee: “De IPython.display importamos display”.
# Para qué sirve: muestra el DataFrame de forma legible dentro del notebook.
from IPython.display import display

# Cómo se lee: “Le pedimos a pandas mostrar hasta veinte columnas cuando imprima una tabla”.
# Para qué sirve: evita que Jupyter oculte columnas del dataset durante la revisión.
pd.set_option("display.max_columns", 20)

# Cómo se lee: “Mostramos un mensaje indicando que las importaciones quedaron listas”.
# Para qué sirve: deja una evidencia sencilla de que esta celda terminó sin errores.
print("Importaciones preparadas correctamente.")
```

## 2. Identificación exacta del dataset

La actividad exige registrar el identificador, la configuración y el split. Se mantienen como variables separadas para que la procedencia quede visible y sea fácil de reutilizar.

**Nota de continuidad:** estas tres constantes son las mismas usadas en EA1. La diferencia es que AE2 conserva las 13 columnas originales para el reconocimiento, mientras EA1 creó dos columnas derivadas y llegó a 15 después de sus transformaciones.


```python
# Cómo se lee: “Guardamos en DATASET_ID el identificador del repositorio de Hugging Face”.
# Para qué sirve: fija la fuente exacta que debe cargar el notebook.
DATASET_ID = "mstz/heart_failure"

# Cómo se lee: “Guardamos en CONFIG la configuración llamada death”.
# Para qué sirve: selecciona la configuración utilizada en la Actividad 1 y en este proyecto.
CONFIG = "death"

# Cómo se lee: “Guardamos en SPLIT la partición train”.
# Para qué sirve: indica que utilizaremos los 299 registros disponibles en ese split.
SPLIT = "train"

# Cómo se lee: “Mostramos en pantalla el identificador, la configuración y el split”.
# Para qué sirve: deja trazabilidad visible antes de descargar los datos.
print(f"Dataset: {DATASET_ID} | Configuración: {CONFIG} | Split: {SPLIT}")
```

## 3. Obtención desde Hugging Face y conversión a DataFrame

La descarga se hace mediante código, como exige la rúbrica. No se usa una copia manual del CSV para construir el DataFrame de esta actividad.

```python
# Cómo se lee: “Cargamos desde Hugging Face el dataset indicado, con la configuración y el split definidos”.
# Para qué sirve: obtiene los datos directamente desde su repositorio de origen para esta actividad.
dataset_hf = load_dataset(DATASET_ID, CONFIG, split=SPLIT)

# Cómo se lee: “Convertimos el objeto de Hugging Face en un DataFrame de pandas”.
# Para qué sirve: deja los datos en una estructura tabular que pandas puede revisar fácilmente.
df = dataset_hf.to_pandas()

# Cómo se lee: “Mostramos el tipo de objeto que quedó almacenado en df”.
# Para qué sirve: verifica que efectivamente trabajamos con un DataFrame de pandas.
print("Tipo de objeto:", type(df))

# Cómo se lee: “Mostramos las dimensiones del DataFrame”.
# Para qué sirve: confirma cuántas filas y cuántas columnas fueron cargadas.
print("Dimensiones:", df.shape)
```

## 4. Algunos registros

Cada fila representa un **registro individual de un paciente con insuficiencia cardíaca** dentro del conjunto documentado. La tabla siguiente permite reconocer visualmente cómo están organizadas las variables.

```python
# Cómo se lee: “Tomamos las primeras cinco filas del DataFrame y las mostramos”.
# Para qué sirve: permite reconocer ejemplos reales de registros y comprobar la organización de las columnas.
display(df.head())
```

## 7. Valores nulos por columna

La actividad solicita identificar cuántos valores faltantes aparecen en cada variable. `isna()` marca las celdas ausentes y `sum()` cuenta esas marcas por columna.

```python
# Cómo se lee: “Preguntamos qué celdas están vacías o son nulas y sumamos el resultado por columna”.
# Para qué sirve: calcula de manera automática la cantidad de valores faltantes de cada variable.
nulos_por_columna = df.isna().sum()

# Cómo se lee: “Mostramos el conteo de valores nulos por columna”.
# Para qué sirve: deja visible la revisión de completitud solicitada en la rúbrica.
print(nulos_por_columna)

# Cómo se lee: “Sumamos todos los nulos encontrados en las columnas y convertimos el resultado a entero”.
# Para qué sirve: permite comprobar rápidamente si el DataFrame tiene algún dato faltante en total.
total_nulos = int(nulos_por_columna.sum())

# Cómo se lee: “Mostramos el total de valores nulos del DataFrame”.
# Para qué sirve: resume el control de completitud en una sola cifra.
print("Valores nulos totales:", total_nulos)
```

## 9. Generación automática de `resultados/resumen_dataframe.txt`

Este bloque usa el **mismo DataFrame `df`** revisado arriba. La carpeta se crea automáticamente cuando no existe y el archivo se sobrescribe en cada ejecución, como solicita la actividad.

```python
# Cómo se lee: “Guardamos en ruta_actual la carpeta desde la que se está ejecutando Jupyter”.
# Para qué sirve: permite construir una ruta relativa sin escribir una ubicación personal del computador.
ruta_actual = Path.cwd()

# Cómo se lee: “Preguntamos si Jupyter está ejecutándose desde una carpeta llamada notebooks”.
# Para qué sirve: hace que el código funcione tanto si abrimos el notebook desde su carpeta como desde la raíz del proyecto.
if ruta_actual.name.lower() == "notebooks":
    # Cómo se lee: “Si estamos dentro de notebooks, tomamos como proyecto la carpeta inmediatamente superior”.
    # Para qué sirve: ubica correctamente la raíz EA2_HeartFailure_Pablo_Gomez.
    ruta_proyecto = ruta_actual.parent
else:
    # Cómo se lee: “Si no estamos dentro de notebooks, usamos la carpeta actual como raíz del proyecto”.
    # Para qué sirve: mantiene el mismo flujo cuando VS Code abre Jupyter desde la carpeta principal.
    ruta_proyecto = ruta_actual

# Cómo se lee: “Construimos la ruta de la carpeta resultados dentro del proyecto”.
# Para qué sirve: define dónde debe quedar el resumen exigido por la actividad.
ruta_resultados = ruta_proyecto / "resultados"

# Cómo se lee: “Creamos la carpeta resultados y permitimos que ya exista sin producir un error”.
# Para qué sirve: garantiza que el notebook pueda guardar el archivo en una ejecución nueva.
ruta_resultados.mkdir(parents=True, exist_ok=True)

# Cómo se lee: “Construimos la ruta completa del archivo resumen_dataframe.txt”.
# Para qué sirve: fija un nombre y una ubicación reproducibles dentro del repositorio.
ruta_resumen = ruta_resultados / "resumen_dataframe.txt"

# Cómo se lee: “Creamos una lista con las primeras líneas de información general del resumen”.
# Para qué sirve: reúne la procedencia y las dimensiones antes de agregar el detalle de columnas.
lineas_resumen = [
    "EA2 - RESUMEN AUTOMÁTICO DEL DATAFRAME",
    "=" * 55,
    f"Dataset: {DATASET_ID}",
    f"Configuración: {CONFIG}",
    f"Split: {SPLIT}",
    "Muestra utilizada: no; se utiliza el split completo.",
    "Fecha de consulta: 25 de octubre de 2026",
    f"Filas: {df.shape[0]}",
    f"Columnas: {df.shape[1]}",
    "",
    "COLUMNAS, TIPOS Y VALORES NULOS",
    "-" * 55,
]

# Cómo se lee: “Recorremos uno por uno los nombres de las columnas del DataFrame”.
# Para qué sirve: permite construir automáticamente una línea de resumen para cada variable existente.
for columna in df.columns:
    # Cómo se lee: “Agregamos el nombre de la columna, su tipo de dato y su cantidad de nulos”.
    # Para qué sirve: completa el detalle solicitado sin escribir manualmente las variables.
    lineas_resumen.append(
        f"{columna} | tipo: {df[columna].dtype} | nulos: {int(df[columna].isna().sum())}"
    )

# Cómo se lee: “Unimos todas las líneas usando un salto de línea entre cada una”.
# Para qué sirve: convierte la lista en un solo texto listo para guardar.
contenido_resumen = "\n".join(lineas_resumen)

# Cómo se lee: “Guardamos el texto en resumen_dataframe.txt usando codificación UTF-8”.
# Para qué sirve: crea o regenera el archivo exigido por AE2 con una codificación estándar.
ruta_resumen.write_text(contenido_resumen, encoding="utf-8")

# Cómo se lee: “Mostramos la ruta absoluta donde quedó guardado el resumen”.
# Para qué sirve: permite verificar visualmente que el archivo se creó en la carpeta correcta.
print("Resumen guardado en:", ruta_resumen.resolve())

# Cómo se lee: “Mostramos una línea en blanco”.
# Para qué sirve: separa visualmente la ruta del contenido del archivo.
print()

# Cómo se lee: “Mostramos en pantalla exactamente el texto que se guardó”.
# Para qué sirve: verifica el contenido generado sin necesidad de abrir otro archivo.
print(contenido_resumen)
```

## 10. Verificación final

Estas comprobaciones no transforman los datos. Solo confirman que el objeto sigue siendo un DataFrame, que corresponde al tamaño esperado y que el resumen quedó disponible.

```python
# Cómo se lee: “Comprobamos que df siga siendo un DataFrame de pandas”.
# Para qué sirve: detiene la ejecución si por error el objeto dejó de ser la tabla esperada.
assert isinstance(df, pd.DataFrame)

# Cómo se lee: “Comprobamos que el DataFrame tenga 299 filas y 13 columnas”.
# Para qué sirve: verifica que se cargó el split completo utilizado en el proyecto.
assert df.shape == (299, 13)

# Cómo se lee: “Comprobamos que el archivo resumen exista en la ruta calculada”.
# Para qué sirve: confirma que la evidencia .txt fue creada correctamente.
assert ruta_resumen.exists()

# Cómo se lee: “Mostramos un mensaje final de validación”.
# Para qué sirve: deja una evidencia clara de que las tres comprobaciones se cumplieron.
print("Validación final completada: DataFrame y resumen .txt disponibles.")
```
